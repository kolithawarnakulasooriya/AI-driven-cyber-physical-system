## 10. Modules

## 12. Decorators
A decorator wraps a function to extend or modify its behavior without changing its source code.

### Example:
```python
def my_decorator(func):
    def wrapper():
        print("Before execution")
        func()
        print("After execution")
    return wrapper
```
